```python
# 1
import pandas as pd

categories = ["billing", "account", "general"]
last_2_digits = [0, 6]

entries = [
    {
        "question" : "what is the annual fee",
        "answer" : "the annual fee is Rs 500.",
        "keywords" : ["fee", "cost", "price", "charge"],
        "category" : "billing"
    },

    {
        "question" : "how to reset password",
        "answer" : "go to setting > reset password",
        "keywords" : ["password", "reset", "login"],
        "category" : "account"
    },

    {
        "question" : "what are your working hours",
        "answer" : "we are open 9 AM to 5 PM.",
        "keywords" : ["hours", "timing", "open", "time"],
        "category" : "general"
    },

    {
        "question" : "how can i pay the fee",
        "answer" : "you can pay via upi or net banking",
        "keywords" : ["pay", "payment", "upi", "fee"],
        "category" : "billing"
    }
  ]
for d in last_2_digits:
  category = categories[d % 3]
  if d==0:
    entry ={
        "question" : "how do I update my registered mobile number",
        "answer" : "have to go to settings > personal info > mobile no and press change",
        "keywords" : ["mobile number", "registered", "contect number", "update"],
        "category" : category
    }
  else:
    entry = {
        "question" : "what happens if i do not pay",
        "answer" : "you will not be able to take part in things and if you want to learn you better pay",
        "keywords" : ["not", "cant", "dont", "wont"],
        "category" : category
    }
  entries.append(entry)
df = pd.DataFrame(entries)
# print(df.to_string(index = False))
# print("no. of entries: ", len(df))
# print("Shape: ", df.shape)


# 2
def score_query(query, df):
    query_words = set(query.lower().split())
    if not query_words:
      return[]
    results = []

    for _, row in df.iterrows():
      keyword_words = set()
      for keyword in row["keywords"]:
        keyword_words.update(keyword.lower().split())
      matched = query_words.intersection(keyword_words)
      if matched:
        score = len(matched)/len(query_words)

        results.append({
            "question": row["question"],
            "answer" : row["answer"],
            "category" : row["category"],
            "score" : score
        })
    results.sort(key=lambda item: item["score"], reverse=True)
    return results

    query = "password reset"

    results = score_query(query, df)
    for item in results:
      print("Question: ", item["question"])
      print("Answer: ", item["answer"])
      print("Category: ", item["category"])
      print("Score: ", round(item["score"],2))
      print()

    results=score_query("fee", df)
    for item in results:
      print(item["question"], item["score"])





#   # 3
#   def same_category(category_name, df):
#     result = df[df["category"] == category_name]
#     return result["question"].tolist()

#   category_name = df.loc[4, "category"]
#   questions = same_category(category_name, df)
#   print("Category:", category_name)
#   print("Questions: ", questions)


#   # 4
#   entry_index=0
#   new_keyword = input("Enter a new keyword: ").strip().lower()
#   if new_keyword:
#     df.at[entry_index, "keywords"].append(new_keyword)
#   print("updated entry: ")
#   print(df.loc[entry_index])
#   roll_number = input("Enter your roll number: ").strip()
#   filename = f"{roll_number}_faq_data.csv"
#   df.to_csv(filename, index=False)
#   print("data is saved to: ", filename)


  # 5
# category_counts = df.groupby("category").size()
# print(category_counts)


# 6
    if not results:
      print("No m atching FAQ found.")
      return []

    highest_score = max(item["score"] for item in results)

    best_matches = [item for item in results
                if item["score"] == highest_score
    ]
    if len(best_matches) > 1:
      print("Tie detected. all best matching FAQs:")
    else:
      print("best matching FAQ:")
    for item in best_matches:
      print("question:", item["question"])
      print("answer:", item["answer"])
      print("category:", item["category"])
      print("score", round(item["score"], 2))
      print()
    return best_matches

    score_query("fee", df)
    score_query("password", df)
```
