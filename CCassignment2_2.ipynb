```python

# Set the random seed to your own roll number using random.seed(your_roll_number) before generating
# your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed
# produces a different list, so this cannot be copied from a classmate).
import random
rollno = 1024170206
random.seed(rollno)
# Generate a list of 100 random numbers between 100 and 900 (inclusive). import random
xs = [random.randint(100,900)for _ in range(100)]
print(xs)

# Count and print all odd numbers in the list.
odd_xs = [x for x in xs if x % 2 != 0]
total_odd_xs = len(odd_xs)
print(total_odd_xs)
print("this is odd no list-- ")
print(odd_xs)

# Count and print all even numbers in the list.
even_xs = [x for x in xs if x % 2 == 0]
total_even_xs = len(even_xs)
print(total_even_xs)
print("this is even no list-- ")
print(even_xs)

# Count and print all prime numbers in the list, and additionally build the actual list of prime numbers
# found using list comprehension.
def isprime(x):
  if x < 2:
    return False

  for i in range(2, x):
    if x % i == 0:
      return False
  return True

prime_xs = [x for x in xs if isprime(x)]
total_prime_xs = len(prime_xs)
print(total_prime_xs)
print("this is prime no list -- ")
print(prime_xs)

# Print the number that occurs most frequently in your list, and how many times it occurs.
most_freq = max(xs, key=xs.count)
times = xs.count(most_freq)
print(most_freq)
print(times)
```

```python
#  Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
# each digit} and set B = {digit × 9 for each digit}. Print both sets.
roll = {1, 0, 2, 4, 1, 7, 0, 2, 0, 6}
A = {x * 7 for x in roll}
print(A)
B = {x * 9 for x in roll}
print(B)
# Find the union of A and B — all unique values across both sets.
print(A|B)
# Find the intersection of A and B — values common to both.
print(A&B)
#  Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
# differs from symmetric_difference().
print(A.difference(B))
print(B.difference(A))
# symmetric difference gives only A and only B ((A-B) + (B-A))
# Find the symmetric difference of A and B.
print(A.symmetric_difference(B))
# Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().
if A.issubset(B):
  print("A is a subset of B")
else:
  print("A is not a subset of B")
print(A.issubset(B))
if B.issuperset(A):
  print("B is a superset of A")
else:
  print("B is not a superset of A")
print(B.issuperset(A))

# Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
# in one line why discard() is safer than remove() when you're not sure the value exists.
x = int(input("Enter a number into a set to check- "))
roll.discard(x)
print(roll)
# discard is safer than remove() cuz it doesnt throw an error if we doesnt have x
```
