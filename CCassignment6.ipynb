```python
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])
total_steps = np.sum(steps)
print(total_steps)
print("mean no of steps:", np.mean(steps))
print("maximum no of steps:", np.max(steps))
print("minimum no of steps:", np.min(steps))
allusers_perdaysteps = steps.sum(axis=0)
alldays_perusersteps = steps.sum(axis=1)
print("each user steps of 3 days:", allusers_perdaysteps)
print("each user 3 days total steps:", alldays_perusersteps)
rowpos, colpos = np.unravel_index(np.argmax(steps), steps.shape)
print("index of user having max steps:", rowpos, colpos)
```

```python
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])
y = np.array([40, 65, 30, 85])
print("shape and dimension of X are", X.shape,"and", X.ndim)
transpose_X = X.T
print("Transpose of X:\n", transpose_X)
product_X = transpose_X @ X
print("product of transpose and X:\n", product_X)
inverse_X = np.linalg.inv(product_X)
print("inverse of product of transpose and X:\n", inverse_X)
transpose_X_y = transpose_X @ y
ols = inverse_X @ transpose_X_y
print("ordinary least square formula: ", ols)
new_user = np.array([5, 40, 7])
prediction = new_user @ ols
print("predicted assistance score-", prediction)
```
