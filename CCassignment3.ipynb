```python
import pandas as pd
employees = {
    "Employee_ID" : [101, 102, 103, 104, 105],
    "Name" : ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department" : ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age" : [29, 34, 41, 28, 38],
    "Salary" : [50000, 70000, 65000, 55000, 60000],
    "Years of experience" : [4, 8, 10, 3, 12],
    "Joining_Date" : ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10 ",  "2010-11-25 "],
    "Gender" : ["Female", "Male", "Male", "Female", "Male"],
    "Bonus" : [5000, 7000, 6000, 4500, 5000],
    "Rating" : [4.5, 4.0, 3.8, 4.7, 3.5],
}
table = pd.DataFrame(employees)
table.to_csv("employees.csv", index = False)

# Shape (number of rows and columns) of the DataFrame.
print(table.shape)
print(table)

# Summary of the DataFrame that includes the data types and non-null counts for
# each column.
table.info()

#  Generate descriptive statistics.
print(table.describe(include = "all"))

# Display the first 5 rows and last 3 rows of the dataset.
print(table.head())
print(table.tail(3))

# Calculate the following statistics from the dataset:
# i. The average salary of employees.
# ii. The total bonus paid to all employees.
# iii. The youngest employee's age.
# iv. The highest performance rating.
print(table["Salary"].mean())
print(table["Bonus"].sum())
print(table["Age"].min())
print(table["Rating"].max())

# Sort the DataFrame by the Salary column in descending order.
print(table.sort_values("Salary", ascending= False))

# Add a new column that categorizes employees based on their performance rating:
# i. Excellent for ratings >= 4.5
# ii. Good for ratings >= 4.0 but < 4.5
# iii. Average for ratings < 4.0
import numpy as np
table["Performance"] = np.select(
    [
        table["Rating"] >= 4.5,
        (table["Rating"] >= 4.0) & (table["Rating"] < 4.5),
        table["Rating"] < 4.0
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default= "Unknown"
)
print(table)

# Identify missing values in the DataFrame.
print(table.isnull().sum())

# Rename the Employee_ID column to ID.
table.rename(columns = {"Employee_ID": "ID"}, inplace= "True")
print(table)

# Find all employees who:
# i. Have more than 5 years of experience.
# ii. Belong to the IT department.
result = table[
    (table["Years of experience"] > 5) &
    (table["Department"] == "IT")
]
print(result)

# Modify the dataset by adding a new column, Tax, which deducts 10% of the
# Salary.
table["Tax"] = table["Salary"] * 0.10
print(table)

# Save the modified DataFrame (with added columns) to a new CSV file.
table.to_csv("updated_employees.csv", index = False)
print("updated_emplyees.csv")
```
